### Intermediate Loops & Loop Control


```python
# 1. Print 1–100 but skip numbers divisible by 5

# continue means skip this number and go to the next one.

for i in range(1, 101):

    if i % 5 == 0:
        continue

    print(i)
```

```python
#2. Stop at the first number divisible by both 7 and 11

#A number divisible by both means:


for i in range(1, 101):

    if i % 7 == 0 and i % 11 == 0:
        print("Found:", i)
        break
```

```python
# 3. Search for a number using for-else
numbers = [10, 20, 30, 40, 50]

search = int(input("Enter a number: "))

for number in numbers:

    if number == search:
        print("Number Found")
        break

else:
    print("Number Not Found")
```

```python
# 4. Use enumerate()
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]

for number, name in enumerate(names, start=1):
    print(number, name)
```

### 7. Multiplication tables from 1 to 10

```python
for i in range(1, 11):

    print("Table of", i)

    for j in range(1, 11):

        print(i, "x", j, "=", i * j)

    print()
```

### 9. Remove duplicates without using set()

```python
numbers = [10, 20, 10, 30, 20, 40, 30]

unique = []

for number in numbers:

    if number not in unique:
        unique.append(number)

print(unique)
```

### 10. Count positive, negative and zero

```python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]

positive = 0
negative = 0
zero = 0

for number in numbers:

    if number > 0:
        positive += 1

    elif number < 0:
        negative += 1

    else:
        zero += 1

print("Positive:", positive)
print("Negative:", negative)
print("Zeros:", zero)
```

### 11. Check whether a number is prime

```python
number = int(input("Enter a number: "))

if number < 2:
    print("Not Prime")

else:

    is_prime = True

    for i in range(2, number):

        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not Prime")
```

### 12. Print all prime numbers from 1 to 100

```python
for number in range(2, 101):

    is_prime = True

    for i in range(2, number):

        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        print(number)
```

```python
Python	    Simple meaning
for	        Repeat something
range()	    Generate numbers
break	    Stop the loop
continue	Skip current turn
else	    Run when condition/loop finishes normally
enumerate()	Get index/number + value
Nested loop	A loop inside another loop
%	        Find remainder
+= 1	    Add 1
append()	Add something to a list
```
