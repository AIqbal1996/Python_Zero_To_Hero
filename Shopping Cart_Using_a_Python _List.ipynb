```python
# Create a cart
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
# 1. Display all products
print("All products:", cart)
```

```python
# 2. Access first and last products
print("First product:", cart[0])
print("Last product:", cart[-1])
```

```python
# 3. Add "Webcam"
cart.append("Webcam")
print("After adding Webcam:", cart)
```

```python
# 4. Insert "USB Hub" at index 2
cart.insert(2, "USB Hub")
print("After inserting USB Hub:", cart)
```

```python
# 5. Remove "Mouse"
cart.remove("Mouse")
print("After removing Mouse:", cart)
```

```python
# 6. Remove the last item using pop()
removed_product = cart.pop()
print("Removed product:", removed_product)
print("After pop():", cart)
```

```python
# 7. Find the index of "Monitor"
print("Index of Monitor:", cart.index("Monitor"))
```

```python
# 8. Count occurrences of "Laptop"
print("Laptop count:", cart.count("Laptop"))
```

```python
# 9. Create a copy of the cart
cart_copy = cart.copy()
print("Cart copy:", cart_copy)
```

```python
# 10. Reverse the cart
cart.reverse()
print("After reverse():", cart)
cart[::-1]  # This is another way to reverse the list
```

```python
# 11. Sort the products alphabetically
cart.sort()
print("After sort():", cart)
```
