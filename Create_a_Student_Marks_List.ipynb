### Create a Student Marks List

```python
marks = [78, 85, 90, 67, 88, 92, 76]
```

```python
# Print complete list
print(marks)
```

```python
type(marks)
```

```python
# Print first element
print(marks[0])
```

```python
# Print last element
print(marks[-1])
```

```python
# Print elements from index 2 to 5
print(marks[2:6])
```

```python
# Number of elements
print(len(marks))
```

```python
# Maximum marks
print(max(marks))
```

```python
# Minimum marks
print(min(marks))
```

```python
# Total marks
print(sum(marks))
```

```python
# Sort in ascending order
marks.sort()
print(marks)
```

```python
# Sort in descending order
marks.sort(reverse=True)
print(marks)
```

```python
# Add 95
marks.append(95)
print(marks)
```

```python
# Add [81, 84]
marks.extend([81, 84])
print(marks)
```

```python
# Remove 67
marks.remove(67)
print(marks)
```

```python
# Count how many times 90 occurs
print(marks.count(90))
```

```python
# Find index of 88
print(marks.index(88))
```

```python
| Code                       | Meaning                 |
| -------------------------- | ----------------------- |
| `marks[0]`                 | First item              |
| `marks[-1]`                | Last item               |
| `marks[2:6]`               | Items from index 2 to 5 |
| `len(marks)`               | Number of items         |
| `max(marks)`               | Highest mark            |
| `min(marks)`               | Lowest mark             |
| `sum(marks)`               | Total marks             |
| `marks.sort()`             | Small → Big             |
| `marks.sort(reverse=True)` | Big → Small             |
| `marks.append(95)`         | Add one item            |
| `marks.extend([81, 84])`   | Add multiple items      |
| `marks.remove(67)`         | Remove 67               |
| `marks.count(90)`          | Count 90                |
| `marks.index(88)`          | Position of 88          |
```
