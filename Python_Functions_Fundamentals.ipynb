## Python Functions Fundamentals

##### Python Functions – Simple Explanation
##### First: What is a Function?

##### A function is a block of code that does one specific job.

```python
def add(a, b):
    return a + b
```

def → creates a function
add → function name
a, b → inputs/parameters
return → sends the answer back

```python
# 1. Add Two Numbers
```

```python
def add(a, b):
    return a + b


print(add(15, 20))
```

#### 2. Addition, Subtraction, Multiplication, Division

```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    return a / b


print(add(10, 5))
print(subtract(10, 5))
print(multiply(10, 5))
print(divide(10, 5))
```

#### 3. Even or Odd

```python
def even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


print(even_odd(11))
print(even_odd(9))
```

#### 4. Largest of Three Numbers

```python
def largest(a, b, c):

    largest_number = a

    if b > largest_number:
        largest_number = b

    if c > largest_number:
        largest_number = c

    return largest_number


print(largest(10, 25, 15))
```

#### 5. Factorial

```python
def factorial(n):

    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result


print(factorial(5))
```

#### 6. Check Prime Number

```python
def is_prime(number):

    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


print(is_prime(7))
print(is_prime(10))
```

#### 7. Default Argument — Discount

```python
def calculate_discount(price, discount=10):

    discount_amount = price * discount / 100

    final_price = price - discount_amount

    return final_price


print(calculate_discount(1000))
```

#### 8. Sum of List Without sum()

```python
def list_sum(numbers):

    total = 0

    for number in numbers:
        total = total + number

    return total


numbers = [10, 20, 30, 40]

print(list_sum(numbers))
```

#### 9. Count Vowels

```python
def count_vowels(text):

    count = 0

    for letter in text:

        if letter.lower() in "aeiou":
            count = count + 1

    return count


print(count_vowels("Hello World"))
```

#### 10. Palindrome

```python
def is_palindrome(text):

    return text == text[::-1]


print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

#### 11. Student Profile

```python
def student_profile(name, age, course):

    return f"Name: {name}, Age: {age}, Course: {course}"
```

```python
# Keyword Arguments

# Here we specify the names:

print(student_profile(
    course="Python",
    name="John Doe",
    age=30
))
```

#### 12. print() vs return

```python
def add_print(a, b):

    print(a + b)


add_print(10, 20)


#But the function doesn't give the value back to us.
```

```python
# Function using return
def add_return(a, b):

    return a + b


result = add_return(10, 20)

print(result)
```
