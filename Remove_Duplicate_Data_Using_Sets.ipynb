Remove Duplicate Data Using Sets

```python
# Create a list
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
# 1. Print the original list
print("Original list:", numbers)
```

```python
# 2. Convert list into a set
unique_numbers = set(numbers)
print("Set:", unique_numbers)
```

```python
# 3. Observe which duplicates disappear
print("Duplicates removed:", set(numbers))
```

```python
# 4. Convert set back into a list
unique_list = list(unique_numbers)
print("Unique list:", unique_list)
```

```python
# 5. Number of original elements
print("Original elements:", len(numbers))
```

```python
# 6. Number of unique elements
print("Unique elements:", len(unique_numbers))
```

```python
# List containing duplicate student names
students = [
    "Rahul",
    "Priya",
    "Aman",
    "Rahul",
    "Neha",
    "Priya",
    "Aman",
    "Rohit"
]
```

```python
# Original list
print("All students:", students)
```

```python
# Convert to set
unique_students = set(students)
print("Unique students:", unique_students)
```

```python
# Convert back to list
unique_student_list = list(unique_students)
print("Unique student list:", unique_student_list)
```

```python
# Count students
print("Total entries:", len(students))
print("Unique students:", len(unique_students))
```

```python
# how many different students
unique_students = set(students)
```
