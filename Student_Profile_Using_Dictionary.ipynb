# Student Profile Using Dictionary


```python
# Create student dictionary
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}
```

```python
# Print complete dictionary
print(student)
```

```python
# Print student's name
print(student["name"])
```

```python
# Print course
print(student["course"])
```

```python
# Print all values
print(student.values())
```

```python
# Print all key-value pairs
print(student.items())
```

```python
# Change marks
student["marks"] = 92
```

```python
# Add email
student["email"] = "rahul@gmail.com"
```

```python
# Add phone
student["phone"] = "123546"
```

```python
# Remove city
student.pop("city")
```

```python
# Use get() to retrieve name
print(student.get("name"))
```

```python
# Create a copy
student_copy = student.copy()
```

```python
# Print updated dictionary
print(student)
```

```python
# Print copied dictionary
print(student_copy)
```

```python
# Dictionary methods in simple words
# 1. keys()

# Gives you all the keys.

print(student.keys())
```

```python
#2. values()

#Gives you all the values.

print(student.values())
```

```python
#3. items()

# Gives you all key-value pairs.

print(student.items())
```

```python
# 4. get()

# Used to get a value safely.

print(student.get("name"))
```

```python
#If the key doesn't exist, get() normally returns None instead of giving an error.

print(student.get("email"))
```

```python
#5. update()

#Used to add or change multiple values.

student.update({
    "marks": 95,
    "city": "Delhi"
})
```

```python
#6. pop()

#Used to remove an item.

student.pop("city")
```

```python
#7. copy()

# Used to make a copy of the dictionary.

student_copy = student.copy()
```

```python
Easy way to remember
keys()   → What are the keys?
values() → What are the values?
items()  → Give me key + value
get()    → Give me this value
update() → Add/change data
pop()    → Remove data
copy()   → Make a copy
```
