### Python for Loop Fundamentals


### 1. Print numbers from 1 to 100

```python
for i in range(1, 101):
    print(i)
```

### 2. Print even numbers from 1 to 100

```python
for i in range(1, 101):

    if i % 2 == 0:
        print(i)
```

### 3. Print odd numbers from 1 to 100

```python
for i in range(1, 101):

    if i % 2 != 0:
        print(i)
```

### 4. Multiplication table from 1 to 20

```python
n = int(input("Enter a number: "))

for i in range(1, 21):

    print(n, "x", i, "=", n * i)
```

### 5. Sum of numbers from 1 to n

```python
n = int(input("Enter a number: "))

total = 0

for i in range(1, n + 1):

    total = total + i

print("Sum:", total)
```

### 6. Factorial of a number

```python
n = int(input("Enter a number: "))

factorial = 1

for i in range(1, n + 1):

    factorial = factorial * i

print("Factorial:", factorial)

```

### 7. Print numbers divisible by 3

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]

for number in numbers:

    if number % 3 == 0:
        print(number)
```

### 8. Print each language with its length

```python
languages = ["Python", "Java", "C++", "JavaScript", "Go"]

for language in languages:

    print(language, len(language))
```

### 9. Print every dictionary key and value

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

for key, value in student.items():

    print(key, ":", value)
```

### 10. Count vowels in a string

```python
text = input("Enter a string: ")

count = 0

for letter in text:

    if letter in "aeiouAEIOU":
        count = count + 1

print("Vowels:", count)
```

### 11. Reverse a string using a for loop

```python
text = input("Enter a string: ")

reverse = ""

for letter in text:

    reverse = letter + reverse

print("Reverse:", reverse)
```

### 12. Find the largest number without max()

```python
numbers = [12, 45, 7, 89, 23, 56]

largest = numbers[0]

for number in numbers:

    if number > largest:
        largest = number

print("Largest:", largest)
```

```python
for       → repeat something
range()   → create a number range
if        → check something
%         → find remainder
len()     → count items/characters
items()   → get dictionary key + value
append()  → add item to a list
```
