# Build a Product Dictionary

```python
# Create laptop dictionary
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}
```

```python
# Display the complete product
print(laptop)
```

```python
# Print brand
print(laptop["brand"])
```

```python
# Print model
print(laptop["model"])
```

```python
# Print price
print(laptop["price"])
```

```python
# Change price
laptop["price"] = 110000
```

```python
# Add GPU information
laptop["GPU"] = "NVIDIA RTX 4050"
```

```python
# Change RAM
laptop["ram"] = "32GB"
```

```python
# Remove available
del laptop["available"]
```

```python
# Print all keys
print(laptop.keys())
```

```python
# Print all values
print(laptop.values())
```

```python
# Print all items
print(laptop.items())
```

```python
# Display updated laptop
print(laptop)
```
