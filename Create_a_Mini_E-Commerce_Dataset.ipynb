# Create a Mini E-Commerce Dataset


```python
# Create a list of dictionaries
products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]
```

```python
# Print all products
print(products)
```

```python
print(type(products))  # Output: <class 'list'>
```

```python
# Print first product
print(products[0])
```

```python
# Print second product's price
print(products[1]["price"])
```

```python
# Print third product's brand
print(products[2]["brand"])
```

```python
# Change first product's price
products[0]["price"] = 65000
```

```python
# Add rating to the second product
products[1]["rating"] = 4.5
```

```python
# Add another product manually
products.append({
    "name": "Smartwatch",
    "price": 20000,
    "brand": "Garmin"
})
```

```python
# Print final dataset
print(products)
```

```python
#Simple Explanation

#Here we are combining a list + dictionaries.

#1. List

#A list stores multiple products.

products = [
    {...},
    {...},
    {...}
]
```

```python
#2. Dictionary

#Each product is a dictionary.

{
    "name": "Laptop",
    "price": 70000,
    "brand": "Dell"
}
```

```python
#3. Indexing

#Lists use index numbers to find an item.

#Python starts counting from 0:

#Index 0 → Laptop
#Index 1 → Phone
#Index 2 → Tablet

#So:

products[0]
```

```python
#4. Keys

#Keys are the names used inside the dictionary.

#"name"
#"price"
#"brand"
products[0]
```

```python
#5. Values

#Values are the actual information stored against the keys.

#"name"  → "Laptop"
#"price" → 70000
#"brand" → "Dell"

#So:

products[0]["price"]
```

```python
Easy way to remember
List
 ↓
Many products
 ↓
Dictionary
 ↓
Product information
 ↓
Keys → name, price, brand
 ↓
Values → Laptop, 70000, Dell
```
