# Creating a nested list

```python
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]
```

```python
type(students)
```

```python
# 1. Print the complete list
print(students)
```

```python
# 2. Print Rahul's name
print(students[0][0][1:3])
```

```python
# 3. Print Priya's age
print(students[1][1])
```

```python
# 4. Print Aman's course
print(students[2][2])
```

```python
# 5. Print the complete record of Priya
print(students[2])
```

```python
# 6. Change Rahul's course to "AI"
students[0][2] = "AI"
```

```python
# 7. Add another student record manually
students.append(["Neha", 23, "Web Development"])
```

```python
# 8. Print the updated nested list
print(students)
```

Understanding Nested List Indexing

```python
students = [
    ["Rahul", 21, "Python"],           # Index 0
    ["Priya", 22, "Data Science"],     # Index 1
    ["Aman", 20, "Machine Learning"]   # Index 2
]
```

First Index → Select the student

```python
students[0] 
```

```python
students[0]   # Rahul's complete record
students[1]   # Priya's complete record
students[2]   # Aman's complete record
```

Second Index → Select information inside that student's record

```python
students[1][0]   # Rahul
students[1][1]   # 21
students[1][2]   # Python
```

Updated List

```python
[
    ["Rahul", 21, "AI"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"],
    ["Neha", 23, "Web Development"]
]
```
