# Advanced Function Challenges

#### 1. *args — Add Any Number of Values

#### *args allows us to pass many values to a function.

```python
def total(*args):
    return sum(args)


print(total(10, 20))
print(total(10, 20, 30, 40))
```

#### 2. *args — Find Largest Number

```python
def largest(*args):
    return max(args)


print(largest(10, 50, 20, 80, 30))
```

#### 3. **kwargs — Create Profile

#### **kwargs allows us to pass information using key=value.

```python
def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


create_profile(
    name="Arif",
    age=30,
    city="Delhi",
    job="Developer"
)
```

#### 4. Function as an Argument

#### One function can be passed to another function.

```python
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


def calculate(operation, a, b):
    return operation(a, b)


print(calculate(add, 10, 20))
print(calculate(multiply, 10, 20))
```

#### 5. Lambda — Square of a Number

#### Lambda is a small one-line function.

```python
square = lambda x: x * x

print(square(5))
```

#### 6. Lambda + map() — Square All Numbers

```python
numbers = [1, 2, 3, 4, 5, 6]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

#### 7. Lambda + filter() — Find Even Numbers

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

#### 8. Recursion — Factorial

#### A function calling itself is called recursion.

```python
def factorial(n):
    if n == 1:
        return 1

    return n * factorial(n - 1)


print(factorial(4))
```

#### 9. Recursion — Sum from 1 to n

```python
def sum_numbers(n):
    if n == 1:
        return 1

    return n + sum_numbers(n - 1)


print(sum_numbers(4))
```

#### 10. Recursion — Fibonacci

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(8))
```

#### 11. Local and Global Variables
#### Global variable

#### A variable created outside a function is global.

#### Local variable

#### A variable created inside a function is local.

```python
x = 10       # Global variable


def test():
    y = 20   # Local variable

    print("Inside:", x)
    print("Inside:", y)


test()

print("Outside:", x)
```

#### 12. Mini Calculator

```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    return a / b


def calculator():
    a = 10
    b = 5

    print("Addition:", add(a, b))
    print("Subtraction:", subtract(a, b))
    print("Multiplication:", multiply(a, b))
    print("Division:", divide(a, b))


calculator()
```
