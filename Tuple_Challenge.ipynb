# Tuple Challenge


```python
# Create a tuple
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

```python
# 1. Print the tuple
print("Tuple:", technologies)
```

```python
# 2. Print its type
print("Type:", type(technologies))
```

```python
# 3. Print the first item
print("First item:", technologies[0])
```

```python
# 4. Print the last item
print("Last item:", technologies[-1])
```

```python
# 5. Slice the tuple
print("Sliced tuple:", technologies[1:4])
```

```python
# 6. Count "Python"
print("Python count:", technologies.count("Python"))
```

```python
# 7. Find the index of "C++"
print("Index of C++:", technologies.index("C++"))
```

```python
# 8. Find the tuple length
print("Tuple length:", len(technologies))
```

```python
# 9. Convert tuple into a list
tech_list = list(technologies)
print("List:", tech_list)
```

```python
type(tech_list)
```

```python
# 10. Add "Go" to the list
tech_list.append("Go")
print("After adding Go:", tech_list)
```

```python
# 11. Convert list back into a tuple
technologies = tuple(tech_list)
print("Final tuple:", technologies)
```

```python
# 1. Create a tuple
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

```python
#2. First and last item
print("First item:", technologies[0])
print("Last item:", technologies[-1])
```

```python
# 3. Slicing
technologies[1:4]
```

```python
# 4. Count
print("Python count:", technologies.count("Python"))
```

```python
# 5. Find index
print("Index of C++:", technologies.index("C++"))
```

```python
# 6. Length
print("Tuple length:", len(technologies))
```

```python
# How to add "Go" to a tuple?

# You cannot directly add an item to a tuple.

# ❌ This doesn't work:

technologies.append("Go")
```

```python
# Tuple → List → Add item → Tuple
tech_list = list(technologies)

tech_list.append("Go")

technologies = tuple(tech_list)
```
