#Set Operations Challenge


```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}

java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

```python
# 1. Print both sets
print("Python students:", python_students)
print("Java students:", java_students)
```

```python
# 2. Students learning either Python or Java
either = python_students.union(java_students)
print("Either Python or Java:", either)
```

```python
# 3. Students learning both Python and Java
both = python_students.intersection(java_students)
print("Learning both:", both)
```

```python
# 4. Students learning only Python
only_python = python_students.difference(java_students)
print("Only Python:", only_python)
```

```python
# 5. Students learning only Java
only_java = java_students.difference(python_students)
print("Only Java:", only_java)
```

```python
# 6. Students belonging to exactly one group
exactly_one = python_students.symmetric_difference(java_students)
print("Exactly one group:", exactly_one)
```

```python
# 7. Add a new student
python_students.add("Vikash")
print("After adding Vikash:", python_students)
```

```python
# 8. Remove a student
python_students.remove("Vikash")
print("After remove():", python_students)
```

```python
# 9. Demonstrate discard()
python_students.discard("Unknown")
print("After discard():", python_students)
```

Understanding the Results

```python
#1. union()

#Combines both sets.
python_students.union(java_students)
```

```python
#2. intersection()

#Finds students present in both sets.
python_students.intersection(java_students)
```

```python
#3. difference()

#Finds students who are in one set but not the other.
python_students.difference(java_students)
```

```python
# For only Java:
java_students.difference(python_students)
```

```python
# 4. symmetric_difference()

# Finds students who belong to exactly one group, not both.

python_students.symmetric_difference(java_students)
```

```python
# 5. add()

# Adds one new item to a set.
python_students.add("Vikash")
```

```python
# 6. remove()

# Removes an item from the set.

python_students.remove("Vikash")

```

```python
# Important: If the student doesn't exist, remove() gives an error.
python_students.remove("Unknown")
# KeyError
```

```python
# 7. discard()

# Also removes an item.
python_students.discard("Vikash")
```

```python
# If the student doesn't exist, discard() does not give an error.

python_students.discard("Unknown")
# No error
```
