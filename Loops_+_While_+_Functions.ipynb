##### 1. Student Result Management System

```python
# We take marks, calculate total, percentage, grade, and pass/fail.
def calculate_total(marks):
    total = 0

    for mark in marks:
        total = total + mark

    return total


def calculate_percentage(total, subjects):
    return total / subjects


def get_grade(percentage):

    if percentage >= 80:
        return "A"

    elif percentage >= 60:
        return "B"

    elif percentage >= 40:
        return "C"

    else:
        return "Fail"


marks = []

for i in range(5):
    mark = int(input("Enter marks: "))
    marks.append(mark)


total = calculate_total(marks)
percentage = calculate_percentage(total, 5)
grade = get_grade(percentage)


print("Total:", total)
print("Percentage:", percentage)
print("Grade:", grade)

if percentage >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")
```

## 2. Banking Application

#### Create functions for balance, deposit, withdrawal, etc.

```python
balance = 10000
history = []


def check_balance():
    print("Balance:", balance)


def deposit():
    global balance

    amount = int(input("Enter amount: "))

    balance = balance + amount

    history.append("Deposited " + str(amount))

    print("Deposit successful")


def withdraw():
    global balance

    amount = int(input("Enter amount: "))

    if amount <= balance:
        balance = balance - amount
        history.append("Withdrawn " + str(amount))

        print("Withdrawal successful")

    else:
        print("Insufficient balance")


def show_history():

    for item in history:
        print(item)


while True:

    print("\n1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. History")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        check_balance()

    elif choice == "2":
        deposit()

    elif choice == "3":
        withdraw()

    elif choice == "4":
        show_history()

    elif choice == "5":
        break
```

### 3. Inventory Management

```python
products = []


def add_product():

    name = input("Product name: ")
    price = int(input("Price: "))
    quantity = int(input("Quantity: "))

    product = {
        "name": name,
        "price": price,
        "quantity": quantity
    }

    products.append(product)


def display_products():

    for product in products:
        print(product)


def search_product():

    name = input("Search product: ")

    for product in products:

        if product["name"] == name:
            print(product)


def inventory_value():

    total = 0

    for product in products:
        total = total + product["price"] * product["quantity"]

    print("Inventory Value:", total)
```

#### 5. Number Analysis Tool

```python
def analyze(numbers):

    largest = numbers[0]
    smallest = numbers[0]

    total = 0
    even = 0
    odd = 0
    positive = 0
    negative = 0

    for number in numbers:

        total = total + number

        if number > largest:
            largest = number

        if number < smallest:
            smallest = number

        if number % 2 == 0:
            even = even + 1
        else:
            odd = odd + 1

        if number > 0:
            positive = positive + 1

        elif number < 0:
            negative = negative + 1

    average = total / len(numbers)

    print("Largest:", largest)
    print("Smallest:", smallest)
    print("Total:", total)
    print("Average:", average)
    print("Even:", even)
    print("Odd:", odd)
    print("Positive:", positive)
    print("Negative:", negative)


analyze([10, 20, -5, 7, 8])
```

#### 6. Employee Salary Analyzer

```python
def salary_analysis(salaries):

    total = 0
    highest = salaries[0]
    lowest = salaries[0]

    for salary in salaries:

        total = total + salary

        if salary > highest:
            highest = salary

        if salary < lowest:
            lowest = salary

    average = total / len(salaries)

    print("Total Payroll:", total)
    print("Average Salary:", average)
    print("Highest Salary:", highest)
    print("Lowest Salary:", lowest)

    print("Above Average:")

    for salary in salaries:

        if salary > average:
            print(salary)


salary_analysis([30000, 50000, 40000, 80000])
```

### 8. Password Strength Checker

```python
def check_password(password):

    uppercase = False
    lowercase = False
    number = False
    special = False

    for char in password:

        if char.isupper():
            uppercase = True

        elif char.islower():
            lowercase = True

        elif char.isdigit():
            number = True

        else:
            special = True


    if (
        len(password) >= 8
        and uppercase
        and lowercase
        and number
        and special
    ):
        return "Strong Password"

    else:
        return "Weak Password"


print(check_password("Hello@123"))
```

#### 9. Prime Number Analyzer

```python
def is_prime(number):

    if number < 2:
        return False

    for i in range(2, number):

        if number % i == 0:
            return False

    return True


start = int(input("Start: "))
end = int(input("End: "))


primes = []


for number in range(start, end + 1):

    if is_prime(number):
        primes.append(number)


print("Prime Numbers:", primes)
print("Count:", len(primes))


total = 0

for number in primes:
    total = total + number


print("Sum:", total)


if len(primes) > 0:
    print("Largest Prime:", primes[-1])
```

#### 10. Expense Tracker

```python
expenses = []


def add_expense():

    name = input("Expense name: ")
    amount = int(input("Amount: "))

    expenses.append({
        "name": name,
        "amount": amount
    })


def view_expenses():

    for expense in expenses:
        print(expense)


def total_expense():

    total = 0

    for expense in expenses:
        total = total + expense["amount"]

    print("Total:", total)


while True:

    print("\n1. Add Expense")
    print("2. View Expenses")
    print("3. Total")
    print("4. Exit")

    choice = input("Choice: ")

    if choice == "1":
        add_expense()

    elif choice == "2":
        view_expenses()

    elif choice == "3":
        total_expense()

    elif choice == "4":
        break
```

#### 11. Mini Authentication System

```python
Username = admin
Password = python123
Maximum attempts = 3
```

```python
correct_username = "admin"
correct_password = "python123"

attempts = 3


while attempts > 0:

    username = input("Username: ")
    password = input("Password: ")

    if username == correct_username and password == correct_password:

        print("Login Successful")

        break

    else:

        attempts = attempts - 1

        print("Login Failed")
        print("Attempts left:", attempts)


if attempts == 0:
    print("Too many failed attempts")
```

#### 12. Super30 Python Utility Application

```python
1. Calculator
2. Palindrome Checker
3. Prime Checker
4. Factorial
5. Multiplication Table
6. Exit
```

```python
def calculator():

    a = int(input("First number: "))
    b = int(input("Second number: "))

    print("Addition:", a + b)


def palindrome():

    text = input("Enter text: ")

    if text == text[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")


def prime():

    number = int(input("Enter number: "))

    if number < 2:
        print("Not Prime")
        return

    for i in range(2, number):

        if number % i == 0:
            print("Not Prime")
            return

    print("Prime")


def factorial():

    number = int(input("Enter number: "))

    result = 1

    for i in range(1, number + 1):
        result = result * i

    print("Factorial:", result)


def table():

    number = int(input("Enter number: "))

    for i in range(1, 11):
        print(number, "x", i, "=", number * i)


while True:

    print("\n--- Python Utility ---")

    print("1. Calculator")
    print("2. Palindrome")
    print("3. Prime")
    print("4. Factorial")
    print("5. Multiplication Table")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        calculator()

    elif choice == "2":
        palindrome()

    elif choice == "3":
        prime()

    elif choice == "4":
        factorial()

    elif choice == "5":
        table()

    elif choice == "6":
        print("Goodbye")
        break

    else:
        print("Invalid choice")
```
