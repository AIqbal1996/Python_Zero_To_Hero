# while Loop Mastery


```python
First: What is while?

A while loop repeats code as long as a condition is true.
```

```python
i = 1

while i < 5:
    print(i)
    i = i + 1
```

#### 1. Print 1 to 100

```python
i = 1

while i <= 100:
    print(i)
    i = i + 1
```

#### 2. Print 100 to 1

```python
i = 100

while i >= 1:
    print(i)
    i = i - 1
```

#### 3. Print Even Numbers from 1 to 100

```python
i = 2

while i <= 100:
    print(i)
    i = i + 2
```

### 4. Sum of Digits

```python
number = 5832
total = 0

while number > 0:
    digit = number % 10
    total = total + digit
    number = number // 10

print(total)
```

#### 5. Reverse an Integer

```python
number = 12345
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

print(reverse)
```

#### 6. Count Number of Digits

```python
number = 5832
count = 0

while number > 0:
    count = count + 1
    number = number // 10

print(count)
```

#### 7. Factorial Using while

```python
number = 5
factorial = 1

while number > 0:
    factorial = factorial * number
    number = number - 1

print(factorial)
```

#### 8. Keep Asking for Numbers Until 0

```python
total = 0

while True:

    number = int(input("Enter number: "))

    if number == 0:
        break

    total = total + number

print("Total:", total)
```

#### 9. Password Checker

```python
correct_password = "python123"

while True:

    password = input("Enter password: ")

    if password == correct_password:
        print("Correct password")
        break

    print("Wrong password")
```

#### 10. Guessing Game

```python
secret_number = 7

while True:

    guess = int(input("Guess the number: "))

    if guess == secret_number:
        print("Correct!")
        break

    print("Wrong! Try again.")
```

##### 11. Menu-Driven Calculator

```python
while True:

    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 5:
        print("Goodbye")
        break

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print("Answer:", a + b)

    elif choice == 2:
        print("Answer:", a - b)

    elif choice == 3:
        print("Answer:", a * b)

    elif choice == 4:
        print("Answer:", a / b)

    else:
        print("Invalid choice")
```

### 12. Simple ATM Menu

```python
balance = 10000

while True:

    print("\n--- ATM ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = int(input("Enter deposit amount: "))
        balance = balance + amount
        print("Deposit successful")

    elif choice == 3:
        amount = int(input("Enter withdrawal amount: "))

        if amount <= balance:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Insufficient balance")

    elif choice == 4:
        print("Thank you!")
        break

    else:
        print("Invalid choice")
```
